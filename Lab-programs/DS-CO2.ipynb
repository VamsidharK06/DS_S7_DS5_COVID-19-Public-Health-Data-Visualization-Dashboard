```python
import dash
from dash import dcc, html, Input, Output

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Square Calculator"),

    dcc.Input(
        id='input',
        type='number',
        placeholder='Enter a number'
    ),

    html.Div(
        id='output'
    )
])


@app.callback(
    Output('output', 'children'),
    Input('input', 'value')
)
def calculate_square(value):
    if value is None:
        return "Enter a number"

    return f"Square: {value ** 2}"


if __name__ == "__main__":
    app.run(debug=True)
```
