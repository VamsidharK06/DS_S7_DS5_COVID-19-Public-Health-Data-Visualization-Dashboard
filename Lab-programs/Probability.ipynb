```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
p = 0.3
data = np.random.binomial(1,p,1000)
sns.histplot(data,stat="density",discrete=True)
plt.title(f"Bernoulli distribution (p={p})")
plt.xlabel("Outcome (0 = failure, 1 = success)")
plt.ylabel("Density")
plt.xticks([0,1])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
n = 10
p = 0.5
data = np.random.binomial(n,p,1000)
sns.histplot(data,kde=True,stat="density",bins=n+1)
plt.title(f"Binomial Distribution (n={n},p={p})")
plt.xlabel("Number of successes")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
lam = 4
data = np.random.poisson(lam,1000)
sns.histplot(data,kde=False,stat="density",bins=range(min(data),max(data)+1))
plt.title(f"Poisson Distribution (lambda={lam})")
plt.xlabel("Number of events")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
mu=0
sigma=1
data=np.random.normal(mu,sigma,1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f"Normal Distribution(mu={mu},sigma={sigma})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
low=0
high=10
data=np.random.uniform(low,high,1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f"Uniform Distribution({low},{high})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
lam = 1.5
data=np.random.exponential(1/lam,1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f"Exponential Distribution(lambad = 1.5)")
plt.xlabel("Time")
plt.ylabel("Density")
plt.show()
```
