```python
# Handling Missing Values using Mean and Mode
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age':[25,30,np.nan,40,35], 'Department':['HR','Finance','Finance',np.nan,'IT']})
print("Original Dataset (With Missing Values):")
print(df)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
# Forward/Backward
import pandas as pd
import numpy as np
df = pd.DataFrame({ 'Age':[25,30,np.nan,40,35], 'Department':['HR','Finance','Finance',np.nan,'IT']})
print("Original Dataset (With Missing Values):")
print(df)
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)
df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
# Deletion
import pandas as pd
import numpy as np
df = pd.DataFrame({ "Age":[25,30,np.nan,40,35], "Department":["HR","Finance","IT","Finance","IT"]})
print("Original Dataset")
print(df)
df_drop_rows = df.dropna()
print("After dropping rows:/n",df_drop_rows)
df_drop_cols = df.dropna(axis=1)
print("After dropping columns:/n",df_drop_cols)
```

```python
# Removing Duplicates
import pandas as pd
df = pd.DataFrame({
    'ID':[1,2,2,3,4,4], 
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
})
print("Original Data:\n",df)
df_exact = df.drop_duplicates()
print("Exact Match Removal:\n",df_exact)
```
